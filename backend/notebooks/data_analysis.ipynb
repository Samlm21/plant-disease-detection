# 🌿 Plant Disease Detection — Exploratory Data Analysis (EDA)

This notebook explores the dataset used for plant disease classification, checking class balance, image dimensions, and distribution across training, validation, and test splits.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from collections import Counter

# Set styling
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
```

## 📂 Dataset Directories and Splits

We inspect the counts of processed data in each split (train, validation, test).

```python
data_dir = "../data/processed"
splits = ["train", "val", "test"]

class_counts = {}
for split in splits:
    split_dir = os.path.join(data_dir, split)
    if os.path.exists(split_dir):
        classes = [d for d in os.listdir(split_dir) if os.path.isdir(os.path.join(split_dir, d))]
        class_counts[split] = {
            c: len(os.listdir(os.path.join(split_dir, c))) for c in classes
        }

df_counts = pd.DataFrame(class_counts).fillna(0).astype(int)
df_counts
```

## 📊 Class Balance Visualization

```python
df_counts.plot(kind="barh", stacked=True)
plt.title("Class Distribution across Train, Validation, and Test Splits")
plt.xlabel("Number of Images")
plt.ylabel("Disease Class")
plt.tight_layout()
plt.show()
```

## 🖼️ Leaf Image Analysis (Resolution & Quality)

Let's sample some images and inspect their properties like width, height, and channels.

```python
resolutions = []
sample_images = []
train_dir = os.path.join(data_dir, "train")

if os.path.exists(train_dir):
    for class_name in df_counts.index:
        cls_dir = os.path.join(train_dir, class_name)
        if os.path.exists(cls_dir):
            files = [f for f in os.listdir(cls_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
            if files:
                img_path = os.path.join(cls_dir, files[0])
                with Image.open(img_path) as img:
                    resolutions.append(img.size)
                    sample_images.append((class_name, img_path))

print(f"Sample image resolutions gathered: {Counter(resolutions)}")
```

## 🎨 Sample Image Plotting

```python
fig, axes = plt.subplots(1, min(len(sample_images), 5), figsize=(20, 4))
for i, (class_name, img_path) in enumerate(sample_images[:5]):
    img = Image.open(img_path)
    axes[i].imshow(img)
    axes[i].set_title(class_name.split("___")[-1].replace("_", " "), fontsize=10)
    axes[i].axis("off")
plt.suptitle("Sample Leaf Images from Dataset", fontsize=14, y=1.05)
plt.show()
```
